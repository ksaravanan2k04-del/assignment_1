```python
class expensesharing:
    def __init__(self, friends):
        self.friends = friends
        self.expenses = {friend: 0 for friend in friends}

    def add_expenses(self, payer, amount, participants):
        split_amount = amount / len(participants)
        for participant in participants:
            self.expenses[participant] += split_amount
        self.expenses[payer] -= amount

    def calculate_settlement(self):
        for friend, balance in self.expenses.items():
            if balance > 0:
                print(f"{friend} needs to be reimbursed: rs.{balance:.2f}")
            elif balance < 0:
                print(f"{friend} owes: rs.{-balance:.2f}")
            else:
                print(f"{friend} is settled up.")
```

```python
if __name__ == "__main__":
    friends = ["saro", "ajeesh", "rajan", "ganash ram"]
    expense_sharing = expensesharing(friends)

    # example usage

    expense_sharing.add_expenses("saro", 2000, ["saro", "ajeesh", "rajan", "ganash ram"])     # saro paid for accommodation
    expense_sharing.add_expenses("ganash ram", 1200, ["saro", "ajeesh", "rajan", "ganash ram"])  # ganash ram paid for shopping
    expense_sharing.add_expenses("ajeesh", 800, ["saro", "ajeesh", "rajan", "ganash ram"])    # ajeesh paid for food
    expense_sharing.add_expenses("rajan", 600, [ "saro", "rajan", "ganash ram"])              # rajan paid for travel

    # calculate settlement
    print("\nFinal Settlement:")  
    expense_sharing.calculate_settlement()
```

Transaction 1: Saro paid Rs. 2000 (accommodation), split among 4 people, Rs. 500 each
    Saro:  Rs. 2000 - Rs. 500 = Rs. 1500 (Saro needs to be reimbursed Rs. 1500)
    Ajeesh: Rs. 0 - Rs. 500 = Rs. -500 (negative balance, Ajeesh owes Rs. 500)
    Rajan:  Rs. 0 - Rs. 500 = Rs. -500 (negative balance, Rajan owes Rs. 500)
    Ganash Ram: Rs. 0 - Rs. 500 = Rs. -500 (negative balance, Ganash Ram owes Rs. 500)


Transaction 2: Ganash Ram paid Rs. 1200 (shopping), split among 4 people, Rs. 300 each
    Ganash Ram: Rs. -500 + (1200 - 300) = Rs. 400
    Saro:  Rs. 1500 - Rs. 300 = Rs. 1200
    Ajeesh: Rs. -500 - Rs. 300 = Rs. -800
    Rajan:Rs. -500 - Rs. 300 = Rs. -800

Transaction 3: Ajeesh paid Rs. 800 (food), split among 4 people, Rs. 200 each
    Ajeesh: Rs. -800 + (800 - 200) = Rs. -200
    Saro:  Rs. 1200 - Rs. 200 = Rs. 1000
    Rajan:  Rs. -800 - Rs. 200 = Rs. -1000
    Ganash Ram: Rs. 400 - Rs. 200 = Rs. 200

Transaction 4: Rajan paid Rs. 600 (travel), split among 3 people only (Saro, Rajan, Ganash Ram), Rs. 200 each
    Rajan:  Rs. -1000 + (600 - 200) = Rs. -600
    Saro:  Rs. 1000 - Rs. 200 = Rs. 800
    Ganash Ram: Rs. 200 - Rs. 200 = Rs. 0
    Ajeesh: Rs. -200 (not part of this travel expense, no change)


Final Balance:
    Saro:  Rs. +800 (Saro needs to be reimbursed Rs. 800)
    Ajeesh: Rs. -200 (Ajeesh owes Rs. 200)
    Rajan:  Rs. -600 (Rajan owes Rs. 600)
    Ganash Ram: Rs. 0 (settled up)
